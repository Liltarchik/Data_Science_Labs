```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
seed = 3
rng = np.random.default_rng(seed)

x = np.linspace(0, 20, 100)

a = 0.55
b = 4.0
sigma = 1.2

y_hat = a * x + b
noise = rng.normal(loc=0, scale=sigma, size=x.size)
y_observed = y_hat + noise

#Об'єднання у матрицю (100 рядків, 4 стовпці) та вивід характеристик
matrix = np.column_stack((x, y_hat, noise, y_observed))

print("Матриця (перші 5 рядків):\n", matrix[:5])
print("\nХарактеристики матриці:")
print("shape:", matrix.shape)
print("ndim: ", matrix.ndim)
print("size: ", matrix.size)
print("dtype:", matrix.dtype)
```
